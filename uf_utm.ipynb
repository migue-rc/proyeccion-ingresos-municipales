```python
from pathlib import Path

import pandas as pd
import plotly.express as px

archivo_uf_utm = Path("exo_data/UF_IVP_UTM.xlsx")

utm_ivp_uf_df = (
    pd.read_excel(
        archivo_uf_utm,
        sheet_name="Cuadro",
        header=2,  # Los encabezados reales están en la tercera fila de Excel.
        parse_dates=["Periodo"],
    )
    .rename(
        columns={
            "Periodo": "fecha",
            "1. Unidad de fomento (UF)": "uf",
            "2. Indice de valor promedio (IVP)": "ivp",
            "3. Unidad tributaria mensual (UTM)": "utm",
        }
    )
)

utm_ivp_uf_df
```

```python
import plotly.express as px

fig = px.line(
    utm_ivp_uf_df,
    x="fecha",
    y=["uf", "ivp", "utm"],
    title="Evolución mensual de UF, IVP y UTM",
    labels={
        "fecha": "Mes y año",
        "value": "Valor en pesos",
        "variable": "Indicador",
    },
    color_discrete_map={
        "uf": "#2563EB",
        "ivp": "#F59E0B",
        "utm": "#16A34A",
    },
)

fig.for_each_trace(lambda trace: trace.update(name=trace.name.upper()))
fig.update_traces(hovertemplate="%{y:,.2f} pesos<extra></extra>")
fig.update_xaxes(tickformat="%m-%Y", hoverformat="%m-%Y")
fig.update_layout(
    hovermode="x unified",
    legend_title_text="Indicador",
    yaxis_tickformat=",.0f",
)

fig.show()
```
